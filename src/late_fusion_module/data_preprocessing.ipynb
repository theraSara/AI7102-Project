```python
import pandas as pd
from pathlib import Path

# Folder where your CSV files are stored
data_dir = Path(r"D:/master/AI7102/project/AI7102-Project/data_with_asr")

# List of your CSV files
split_files = ["train_split.csv", "val_split.csv", "test_split.csv"]

# Old and new prefixes
old_prefix = "/Users/vanilla/Documents/courses/AI7102/project/AI7102-Project-1/IEMOCAP_full_release"
new_prefix = r"D:/master/AI7102/project/IEMOCAP_full_release"

for split in split_files:
    file_path = data_dir / split
    print(f"Processing {file_path}...")

    # Load CSV
    df = pd.read_csv(file_path)

    # Replace paths directly in 'audio_path'
    df["audio_path"] = df["audio_path"].str.replace(old_prefix, new_prefix, regex=False)

    # Save back to the same file
    df.to_csv(file_path, index=False)

    print(f"✅ Updated and saved: {file_path}")

print("All splits updated successfully!")
```
